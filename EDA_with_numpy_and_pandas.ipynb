**Find country with the lowest life expetancy**

```python
min_exp = df['life_exp'].min()
df[df['life_exp'] == min_exp]
```

```python
max_exp = df['life_exp'].max()
df[df['life_exp'] == max_exp]
```

```python
df['gdp_cap'].max()
```

```python
df['gdp_cap'].min()
```

**Find Country with the min and max gdp_cap**

```python
min_gdp = df['gdp_cap'].min()
df[df['gdp_cap'] == min_gdp]
```

```python
max_gdp = df['gdp_cap'].max()
df[df['gdp_cap'] == max_gdp]
```

**Total GDP of every continent in 2007**

```python
df_2007 = df[df['year']  == 2007]
df_2007.head()
```

```python
df_2007['gdp'] = df_2007['population'] * df_2007['gdp_cap']
df_2007.head()
```

```python
# Rank countries by gdp
df_2007.sort_values(by='gdp').tail()
```

```python
df_2007.sort_values(by='gdp_cap',ascending=
False).head(10)
```

```python
df_2007.groupby(by='continent')[['gdp','gdp_cap']].sum()
```

**Find average life expectancy of each continent in 2007**

```python
df_2007.groupby(by='continent')['life_exp'].mean()
```
